```python
"""
Print numbers from 1 to 100 with the following rules:
Multiples of 3 print Fizz
Multiples of 5 print Buzz
Multiples of both print FizzBuzz
"""


for number in range (1,100):
    if number%3==0 and number%5==0:
      print("FizzBuzz")
    elif number %3 ==0 :
      print("Fizz")
    elif  number %5 ==0 :
      print("Buzz")
    else:
      print(number)


```
