```python
#0 to 13 KID/13 to 18 minor/18 to 23-teenage/23 to 35-young adult/36-46-middle aged 46-100--no details
age=int(input("enter your age:-"))
print(f"your age is {age}")

if age>=0 and age<13:
  print("KID")
elif age>=13 and age<18:
  print("minor")
elif age>=18 and age<23:
  print("teenage")
elif age>=23 and age<35:
  print("young adult")
elif age>=36 and age<=46:
  print("middle aged")
elif age>=46 and age<=100:
  print("older person")
else:
  print("no details found")



```
